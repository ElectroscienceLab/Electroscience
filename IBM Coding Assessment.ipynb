## Coding (Standard) 1

given an integer n, determine the following:

1. how many 1-bits are in its binary representation?
2. the number *n*'s binary representation has *k* significant bits indexed from 1 to *k*. What are the respective positions of each 1-bit, in ascending order?

*note*: the leftmost 1 bit is always position 1. Preceding zeros are not considered in determining the position.

Function description:
Complete the function *getOneBits* in the editor below. The function must return a results array with the number of 1's stored at *results[0]* followed by the positions of all 1's in its binary representation in ascending order.

*getOneBits* has the following parameter(s): 

n: an integer

Contraints:

1<n<10^9

FINISH THE CODE TO CONFORM INSTRUCTIONS ABOVE:
```
def getOneBits(n)

if __name__== '__main__': ...
```

````python
def getOneBits(n):
    # Convert the number to binary and remove the '0b' prefix
    binary_rep = bin(n)[2:]
    # Initialize the results list with the count of '1's in the binary representation
    results = [binary_rep.count('1')]
    # Add the positions of '1's to the results list
    results.extend([i + 1 for i in range(len(binary_rep)) if binary_rep[i] == '1'])
    return results

if __name__ == '__main__':
    # Example usage
    n = 19
    print(getOneBits(n))  # Output should be [3, 1, 2, 5]
````

This function first converts the integer *n* to its binary representation (excluding the '0b' prefix). It counts the number of '1' bits, and then finds their positions.

## Coding (Standard) 2

The table below contains some reference values for converting between arabic numerals and roman numerals:


```
     Arabic	Roman	 	Arabic	Roman
	   1	  I	 	  40	  XL
	   2	  II	 	  50	  L
	   3 	  III		  90	  XC

Given an integer, convert it to its roman numeral equivalent

**Example:**
numbers = [1,49,23]

Looking at the conversions above, 1 is represented as I (capital), 49 is 40+9 so XLIX and 23 is XXIII. The return array is ['I,'XLIX','XXIII']

**Function Description:**

Complete the function <code>romanizer</code> in the editor below. The function must return as array of strings that represent the integers as their Roman numeral equivalents.

romanizer has the following parameter(s):

<code>int numbers[n]</code>: an array of integers

Returns:

<code>string[n]</code>: an array of strings that represent the integers as their Roman numeral equivalents.

**Contraints:**

1 less than equal n less than equal 1000
1 less than equal <code>numbers[i]</code> less than equal 1000


FINISH THE CODE TO CONFORM INSTRUCTIONS ABOVE:

#write your code here

```

````python
def romanizer(numbers):
    roman_numerals = {
        1000: 'M', 900: 'CM', 500: 'D', 400: 'CD',
        100: 'C', 90: 'XC', 50: 'L', 40: 'XL',
        10: 'X', 9: 'IX', 5: 'V', 4: 'IV',
        1: 'I'
    }

    def int_to_roman(num):
        result = ''
        for value in sorted(roman_numerals.keys(), reverse=True):
            while num >= value:
                result += roman_numerals[value]
                num -= value
        return result

    return [int_to_roman(num) for num in numbers]

if __name__ == '__main__':
    numbers = [1, 49, 23]
    print(romanizer(numbers))  # Output should be ['I', 'XLIX', 'XXIII']
````

This function <code>romanizer</code> converts a list of integers to their Roman numeral equivalents using a dictionary for Roman numeral values. It iterates over each number, constructs the Roman numeral by appending symbols from the largest to smallest value until the number is reduced to zero.

### **SOLUTION:**

````python
def maxXor(lo, hi, k):
    max_xor = 0
    for a in range(lo, hi + 1):
        for b in range(a, hi + 1):
            xor_val = a ^ b
            if xor_val <= k:
                max_xor = max(max_xor, xor_val)
    return max_xor

if __name__ == '__main__':
    import sys
    input = sys.stdin.read
    data = list(map(int, input().strip().split()))
    lo, hi, k = data[0], data[1], data[2]
    print(maxXor(lo, hi, k))
````

An array of integers is defined as being in meandering order when the first two elements are the respective largest and smallest elements in the array and the subsequent elements alternate between its next largest and nextsmallest elements. In other words the elements are in order of <i>[first_largest, first_smallest, second_largest, second_smallest,...]</i>

### **SOLUTION:**

````python
def meanderingArray(unsorted):
    # Sort the input array
    sorted_arr = sorted(unsorted)
    
    # Initialize pointers and result array
    left = 0  # Smallest element pointer
    right = len(sorted_arr) - 1  # Largest element pointer
    result = []
    
    # Alternately add the largest and smallest elements
    while left <= right:
        if right >= left:
            result.append(sorted_arr[right])  # Add the largest remaining element
            right -= 1
        if left <= right:
            result.append(sorted_arr[left])  # Add the smallest remaining element
            left += 1
    
    return result

if __name__ == '__main__':
    n = int(input())  # Read the size of the array
    unsorted = []
    
    for _ in range(n):
        unsorted.append(int(input().strip()))  # Read the elements one by one
    
    result = meanderingArray(unsorted)  # Get the meandering array
    
    # Print each element in the result array on a new line
    for num in result:
        print(num)
````

To match the expected output, we need to ensure that each number in the *meanderingArray* result is printed on a new line instead of printing the entire list at once.

Instead of printing the entire list at once <code>(print(result))</code>, we iterate through each element in the result list and print them individually on a new line using a<code> for loop: for num</code> in <code>result: print(num)</code>.
